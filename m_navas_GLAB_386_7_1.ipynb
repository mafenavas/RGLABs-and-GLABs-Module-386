**Prompts for Descriptive pricing analysis**


- What’s the overall price distribution (median, IQR, tails)?

- Which room types are most/least expensive?

- Are there obvious outliers to cap or investigate?

````python
# Assuming the merged dataset is called airbnb_df

# Convert price column to numeric (if needed)
merged_df['price'] = merged_df['price'].replace('[\$,]', '', regex=True).astype(float)

# 1. Overall price distribution: Median and IQR

median_price = merged_df['price'].median()

Q1 = merged_df['price'].quantile(0.25)
Q3 = merged_df['price'].quantile(0.75)

IQR = Q3 - Q1

print("Price Distribution:")
print(f"Median price: ${median_price:.2f}")
print(f"Q1 (25th percentile): ${Q1:.2f}")
print(f"Q3 (75th percentile): ${Q3:.2f}")
print(f"IQR: ${IQR:.2f}")


# 2. Average price by room type

room_prices = merged_df.groupby('room_type')['price'].mean().sort_values(ascending=False)

print("\nAverage price by room type:")
print(room_prices)


# 3. Detect possible outliers using IQR method

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = merged_df[
    (merged_df['price'] < lower_limit) |
    (merged_df['price'] > upper_limit)
]

print("\nOutlier Analysis:")
print(f"Lower price limit: ${lower_limit:.2f}")
print(f"Upper price limit: ${upper_limit:.2f}")
print(f"Number of possible outliers: {len(outliers)}")
````

**Propmts for Inventory mix & positioning**


- What share of listings are Entire home/apt vs Private room vs others?

 - Do some room types dominate certain price bands?

````python
# 1. Share of listings by room type

room_type_counts = merged_df['room_type'].value_counts()

room_type_share = (room_type_counts / len(merged_df)) * 100

print("Listing share by room type:")
print(room_type_share.round(2))


# 2. Create price bands

bins = [0, 50, 100, 200, 500, float('inf')]
labels = ['<$50', '$50-$100', '$100-$200', '$200-$500', '$500+']

merged_df['price_band'] = pd.cut(
    merged_df['price'],
    bins=bins,
    labels=labels
)


# 3. Analyze room types by price band

price_band_room_type = pd.crosstab(
    merged_df['price_band'],
    merged_df['room_type'],
    normalize='index'
) * 100

print("\nRoom type distribution within each price band (%):")
print(price_band_room_type.round(2))
````

Possible insights:

Private rooms may dominate lower price ranges because they are cheaper options.
Entire homes/apartments often dominate higher price bands due to larger spaces and more amenities.
Shared rooms are usually concentrated in the lowest price categories.

**Propmt for Review recency & activity**

- How recent are reviews? Is demand cooling/heating?

- Are there seasonal peaks in last reviews?

````python
# Convert last_review to datetime
merged_df['last_review'] = pd.to_datetime(merged_df['last_review'])

# 1. Review recency analysis

# Get the most recent review date in the dataset
latest_review = merged_df['last_review'].max()

# Calculate how many days/months ago the last review happened
merged_df['days_since_review'] = (
    latest_review - merged_df['last_review']
).dt.days

print("Review Recency:")
print(f"Most recent review date: {latest_review}")
print(f"Average days since last review: {merged_df['days_since_review'].mean():.0f} days")
print(f"Median days since last review: {merged_df['days_since_review'].median():.0f} days")


# 2. Check demand trend (based on review activity)

reviews_by_month = (
    merged_df['last_review']
    .dt.to_period('M')
    .value_counts()
    .sort_index()
)

print("\nNumber of reviews by month:")
print(reviews_by_month)


# 3. Identify seasonal peaks

reviews_by_month_number = (
    merged_df['last_review']
    .dt.month
    .value_counts()
    .sort_index()
)

print("\nReviews by month (seasonality):")
print(reviews_by_month_number)
````

Possible conclusion:

Summer peaks may indicate higher travel demand.
Holiday months may show seasonal increases.
Lower months may reflect slower travel periods.

**Propmts for Market segmentation**

- Are there price differences across room-type segments?

- If we have neighborhood/borough (not seen here), how does location shift price?

````python
# Convert price column to numeric if needed
merged_df['price'] = merged_df['price'].replace('[\$,]', '', regex=True).astype(float)

# Average price by room type
room_type_prices = (
    merged_df
    .groupby('room_type')['price']
    .agg(['count', 'mean', 'median'])
    .sort_values('mean', ascending=False)
)

print("Price comparison by room type:")
print(room_type_prices)
````

Interpretation:
Entire home/apt is usually the most expensive because guests rent the complete property.
Private Room tends to have mid-range prices.
Shared Room is usually the cheapest option.
